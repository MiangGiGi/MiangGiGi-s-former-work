```python
import numpy as np
import math
from warnings import filterwarnings
filterwarnings('ignore')
```

```python
#定义
population_size = 10  # 种群数量
Crossing_P=0.6 # 交叉概率
Mutant_P=0.01 #变异概率
X_max=5    #最大值
X_min=0     #最小值
DNA_SIZE=10  #DNA长度,当前保留2位小数点
Generations=5000
```

```python
#目标函数
def aim(x):return 11*np.sin(4*x)+9*np.cos(3*x)
```

```python
#解码函数
def decode(population):
   return   population.dot(2 ** np.arange(DNA_SIZE)[::-1]) *(X_max-X_min)/ float(2**DNA_SIZE-1) +X_min
```

```python
#适应度函数
def Fitness(x):
    return x + 1e-3 + abs(np.min(x))
```

```python
#选择函数（轮盘赌）
def select(population, fitness):           
    # print(abs(fitness))
    # print(fitness.sum())
    idx = np.random.choice(np.arange(population_size), size=population_size, replace=True,p=fitness/fitness.sum())
    # print(idx)
    return population[idx]
```

```python
#交叉函数
def Cross(parent, population):
    if np.random.rand() < Crossing_P:    
        i_ = np.random.randint(0, population_size, size=1)
        cross_points = np.random.randint(0, 2, size=DNA_SIZE).astype(np.bool_)
        parent[cross_points] = population[i_, cross_points]
    return parent 
```

```python
#变异函数
def Mutate(child):                 
    for point in range(DNA_SIZE):
        if np.random.rand() < Mutant_P:
            child[point] = 1 if child[point] == 0 else 0
    return child
```

```python
population = np.random.randint(2, size=(population_size, DNA_SIZE))
for i in range(Generations):
    X_value= decode(population)
    F_values = aim(X_value)
    fitness = Fitness(F_values)
    if(i==0):
        max=np.max(F_values)
        max_DNA = population[np.argmax(F_values), :]
    if(max<np.max(F_values)):
        max=np.max(F_values)
        max_DNA=population[np.argmax(F_values), :]
        
    if (i % 10 == 0 and i<200):
        print("第"+str(i)+"代中最值点的Y和X值: \n",
              round(np.max(F_values),2), round(decode(population[np.argmax(F_values), :]),2))
#         print("\n")
    elif(i%100==0):
        print("第"+str(i)+"代中最值点的Y和X值: \n",
              round(np.max(F_values),2), round(decode(population[np.argmax(F_values), :]),2))
#         print("\n")
    #选择
    population = select(population,fitness)
    # print(pop)
    population_copy = population.copy()
    # print(pop_copy)
    for parent in population:
        # print(parent)
        child = Cross(parent,population_copy)
        child =Mutate(child)
        # print(child)
        parent[:] = child
print("目标函数最大值为：",round(max,2))
print("其DNA值为：",max_DNA)
print("其X值为：",round(decode(max_DNA),2))
```
