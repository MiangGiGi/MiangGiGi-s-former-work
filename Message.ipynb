```python
import serial as ser # 导入串口库

se = ser.Serial("/dev/ttyTHS1", 115200, timeout=1) # 设置要使用的引脚、波特率和超时时间

# 实时控制帧数据打包函数
def point_data_pack(down_flag,is_flag,x,y):
    pack_data = bytearray([0xAA,0x29,0x05,0x41,0x00,is_flag,down_flag,x,x>>8,y,y>>8,0x00])
    lens = len(pack_data)
    pack_data[4] = 7;
    i = 0
    sum = 0
    while i<(lens-1):
        sum = sum + pack_data[i]
        i = i+1
        
    pack_data[lens-1] = sum;
    return pack_data

# 角度距离偏差控制数据打包函数
def point_data_pack(is_flag,down_flag,distance,angle):
    pack_data = bytearray([0xAA,0x29,0x05,0x41,0x00,is_flag,down_flag,distance,distance>>8,angle,angle>>8,0x00])
    lens = len(pack_data)
    pack_data[4] = 7;
    i = 0
    sum = 0
    while i<(lens-1):
        sum = sum + pack_data[i]
        i = i+1
    pack_data[lens-1] = sum;
    return pack_data

# 数据发送函数
def send_data(data):
    se.write(data)
```
